### Importando as bibliotecas, exportando os datasets e tratamento de dados

```python
### Importando as bibliotecas

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

plt.rcParams["figure.figsize"] = (12,6)
```

```python

## Carregamentos dos dados

import pandas as pd
customers = pd.read_csv("olist_customers_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
geolocation = pd.read_csv("olist_geolocation_dataset.csv")
```

```python
## Analisando o formato dos dados

datasets = {
    'customers': customers,
    'orders': orders,
    'order_items': order_items,
    'payments': payments,
    'order_reviews': order_reviews,
    'products': products,
    'sellers': sellers
}

for name, df in datasets.items():

    print("\n" + "="*70)
    print(name.upper())
    print(f"Shape: {df.shape}")

    display(df.head())

    print("\nTipos de dados:")
    display(df.dtypes)

    print("\nValores nulos:")
    display(df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
## Convertendo as datas

date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_columns:
    orders[col] = pd.to_datetime(
        orders[col],
        errors='coerce'
    )
```

```python
## Construindo a base unificada

df = (
    orders
    .merge(
        customers[['customer_id','customer_state']],
        on='customer_id',
        how='left'
    )
    .merge(
        order_reviews[['order_id','review_score']],
        on='order_id',
        how='left'
    )
)
```

```python
## Criando as regiões

regioes = {

    'AC':'Norte','AP':'Norte','AM':'Norte',
    'PA':'Norte','RO':'Norte','RR':'Norte','TO':'Norte',

    'AL':'Nordeste','BA':'Nordeste','CE':'Nordeste',
    'MA':'Nordeste','PB':'Nordeste','PE':'Nordeste',
    'PI':'Nordeste','RN':'Nordeste','SE':'Nordeste',

    'DF':'Centro-Oeste','GO':'Centro-Oeste',
    'MT':'Centro-Oeste','MS':'Centro-Oeste',

    'ES':'Sudeste','MG':'Sudeste',
    'RJ':'Sudeste','SP':'Sudeste',

    'PR':'Sul','RS':'Sul','SC':'Sul'
}

df['regiao'] = df['customer_state'].map(regioes)
```

### Satisfação do Cliente por estado

```python
## Satisfação do cliente por estado

satisfacao_estado = (
    df.groupby('customer_state')['review_score']
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

display(satisfacao_estado)
```

### Evolução histórica da satisfação do cliente e volume de avaliações

```python
## Satisfação semestral

# Base para análise semestral
evolucao_satisfacao = df.copy()

# Cria coluna ano_semestre
evolucao_satisfacao['ano'] = evolucao_satisfacao['order_purchase_timestamp'].dt.year
evolucao_satisfacao['semestre'] = evolucao_satisfacao['order_purchase_timestamp'].dt.month.apply(lambda x: 1 if x <= 6 else 2)
evolucao_satisfacao['ano_semestre'] = evolucao_satisfacao['ano'].astype(str) + '-S' + evolucao_satisfacao['semestre'].astype(str)

# Agrupa por semestre
satisfacao_semestral = (
    evolucao_satisfacao
    .groupby('ano_semestre')
    .agg(
        satisfacao_media=('review_score', 'mean'),
        total_avaliacoes=('review_score', 'count')
    )
    .reset_index()
)

# Filtra para o período relevante (mesmo de satisfacao_mensal para consistência)
satisfacao_semestral = satisfacao_semestral[
    (satisfacao_semestral['ano_semestre'] >= '2017-S1') &
    (satisfacao_semestral['ano_semestre'] <= '2018-S1')
]

display(satisfacao_semestral)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Base para análise mensal
satisfacao_mensal = df.copy()

# Cria coluna ano_mes
satisfacao_mensal['ano_mes'] = (
    satisfacao_mensal['order_purchase_timestamp']
    .dt.to_period('M')
    .astype(str)
)

# Agrupa por mês
satisfacao_mensal = (
    satisfacao_mensal
    .groupby('ano_mes')
    .agg(
        satisfacao_media=('review_score', 'mean'),
        total_avaliacoes=('review_score', 'count')
    )
    .reset_index()
)

# Filtra de jan/2017 até jun/2018
satisfacao_mensal = satisfacao_mensal[
    (satisfacao_mensal['ano_mes'] >= '2017-01') &
    (satisfacao_mensal['ano_mes'] <= '2018-06')
]

# Garante índice sequencial
satisfacao_mensal = satisfacao_mensal.reset_index(drop=True)

# Formata datas
satisfacao_mensal['ano_mes_formatado'] = pd.to_datetime(
    satisfacao_mensal['ano_mes']
).dt.strftime('%b/%Y')

# Estilo
sns.set_style("white")
plt.rcParams['axes.grid'] = False

# Figura
fig, ax1 = plt.subplots(figsize=(15, 7))

# ==========================
# BARRAS - SATISFAÇÃO MÉDIA
# ==========================
barras = ax1.bar(
    range(len(satisfacao_mensal)),
    satisfacao_mensal['satisfacao_media'],
    color='#1F4E79',
    alpha=0.9,
    label='Satisfação Média'
)

# Valores nas barras
for barra in barras:

    altura = barra.get_height()

    ax1.annotate(
        f'{altura:.2f}',
        (
            barra.get_x() + barra.get_width()/2,
            altura
        ),
        textcoords='offset points',
        xytext=(0, 5),
        ha='center',
        fontsize=8,
        color='#1F4E79',
        fontweight='bold'
    )

ax1.set_ylabel(
    'Satisfação Média',
    fontsize=12
)

# Ajusta o eixo para destacar variações
ax1.set_ylim(
    satisfacao_mensal['satisfacao_media'].min() - 0.15,
    satisfacao_mensal['satisfacao_media'].max() + 0.15
)

# ==========================
# LINHA - VOLUME DE AVALIAÇÕES
# ==========================
ax2 = ax1.twinx()

linha = ax2.plot(
    range(len(satisfacao_mensal)),
    satisfacao_mensal['total_avaliacoes'],
    color='#8A8A8A',
    marker='o',
    linewidth=2,
    label='Volume de Avaliações'
)

ax2.set_ylabel(
    'Volume de Avaliações',
    fontsize=12
)

# ==========================
# DATAS
# ==========================
ax1.set_xticks(range(len(satisfacao_mensal)))

ax1.set_xticklabels(
    satisfacao_mensal['ano_mes_formatado'],
    rotation=45,
    ha='right'
)

# ==========================
# TÍTULO
# ==========================
plt.title(
    'Evolução da Satisfação dos Clientes e Volume de Avaliações',
    fontsize=16,
    fontweight='bold',
    pad=20
)

# ==========================
# LEGENDA
# ==========================
handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    handles1 + handles2,
    labels1 + labels2,
    loc='upper left',
    frameon=False
)

# ==========================
# LIMPEZA VISUAL
# ==========================
ax1.grid(False)
ax2.grid(False)

sns.despine(right=False)

plt.tight_layout()
plt.show()
```

A satisfação dos clientes manteve-se elevada durante quase todo o período, com médias próximas ou superiores a 4 pontos. Entre novembro de 2017 e março de 2018 houve uma queda na satisfação, coincidindo com um aumento expressivo no volume de avaliações, o que pode indicar desafios operacionais em períodos de maior demanda. A partir de abril de 2018, os índices voltaram a crescer, atingindo o maior nível da série em junho de 2018 (4,28).

De acordo com nossas descobertas, os períodos de menor satisfação foram
associados a aumentos significativos na taxa de atraso e no tempo médio de entrega.
Acreditamos que a deterioração do desempenho logístico foi a razão para a menor
satisfação dos clientes.

```python
total_pedidos = len(df)

pedidos_cancelados = (
    df['order_status'] == 'canceled'
).sum()

taxa_cancelamento = (
    pedidos_cancelados / total_pedidos
) * 100

print(f'Taxa de cancelamento: {taxa_cancelamento:.2f}%')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter

# Base para análise de cancelamento mensal
cancelamento_mensal = orders.copy()

# Cria coluna ano_mes
cancelamento_mensal['ano_mes'] = (
    cancelamento_mensal['order_purchase_timestamp']
    .dt.to_period('M')
    .astype(str)
)

# Identifica pedidos cancelados
cancelamento_mensal['cancelado'] = (
    cancelamento_mensal['order_status'] == 'canceled'
).astype(int)

# Agrupa por mês e calcula a taxa de cancelamento
cancelamento_mensal = (
    cancelamento_mensal
    .groupby('ano_mes')
    ['cancelado']
    .mean()
    .mul(100) # Converte para porcentagem
    .reset_index()
    .rename(columns={'cancelado': 'taxa_cancelamento'})
)

# Filtra de jan/2017 até jun/2018 (período de análise)
cancelamento_mensal = cancelamento_mensal[
    (cancelamento_mensal['ano_mes'] >= '2017-01') &
    (cancelamento_mensal['ano_mes'] <= '2018-06')
].sort_values('ano_mes')

# Garante índice sequencial
cancelamento_mensal = cancelamento_mensal.reset_index(drop=True)

# Formata as datas para exibição
cancelamento_mensal['ano_mes_formatado'] = pd.to_datetime(
    cancelamento_mensal['ano_mes']
).dt.strftime('%b/%Y')

# Estilo
sns.set_style("white")
plt.rcParams['axes.grid'] = False

# Figura
fig, ax = plt.subplots(figsize=(14, 6))

# Cores do relatório
COR_PRINCIPAL = '#1F4E79'
COR_AREA = '#A9C7E8'
COR_MAX = '#2A9D8F'   # Verde
COR_MIN = '#E63946'   # Vermelho
COR_MEDIA = '#6E6E6E'

# Gráfico de linha
sns.lineplot(
    data=cancelamento_mensal,
    x='ano_mes_formatado',
    y='taxa_cancelamento',
    marker='o',
    markersize=7,
    linewidth=3,
    color=COR_PRINCIPAL,
    ax=ax
)

# Área sombreada
ax.fill_between(
    range(len(cancelamento_mensal)),
    cancelamento_mensal['taxa_cancelamento'],
    alpha=0.20,
    color=COR_AREA
)

# Taxa em todos os pontos
for i, valor in enumerate(cancelamento_mensal['taxa_cancelamento']):

    ax.annotate(
        f'{valor:.2f}%',
        (i, valor),
        textcoords='offset points',
        xytext=(0, 8),
        ha='center',
        fontsize=8,
        color=COR_PRINCIPAL
    )

# Linha da média
media = cancelamento_mensal['taxa_cancelamento'].mean()

ax.axhline(
    media,
    linestyle='--',
    linewidth=2,
    color=COR_MEDIA,
    alpha=0.8
)

# Texto da média
ax.text(
    len(cancelamento_mensal) - 0.5,
    media + 0.03,
    f'Média: {media:.2f}%',
    fontsize=9,
    color=COR_MEDIA,
    ha='right'
)

# Máximo e mínimo
max_idx = cancelamento_mensal['taxa_cancelamento'].idxmax()
min_idx = cancelamento_mensal['taxa_cancelamento'].idxmin()

max_y = cancelamento_mensal.loc[max_idx, 'taxa_cancelamento']
min_y = cancelamento_mensal.loc[min_idx, 'taxa_cancelamento']

max_data = cancelamento_mensal.loc[max_idx, 'ano_mes_formatado']
min_data = cancelamento_mensal.loc[min_idx, 'ano_mes_formatado']

# Destaque máximo (verde)
ax.scatter(
    max_idx,
    max_y,
    s=120,
    color=COR_MAX,
    zorder=5,
    edgecolors='white',
    linewidth=2
)

ax.annotate(
    f'Máximo\n{max_data}',
    (max_idx, max_y),
    textcoords='offset points',
    xytext=(0, 35),
    ha='center',
    fontsize=9,
    fontweight='bold',
    color=COR_MAX,
    bbox=dict(
        boxstyle='round,pad=0.3',
        fc='white',
        ec=COR_MAX,
        alpha=0.95
    )
)

# Destaque mínimo (vermelho)
ax.scatter(
    min_idx,
    min_y,
    s=120,
    color=COR_MIN,
    zorder=5,
    edgecolors='white',
    linewidth=2
)

ax.annotate(
    f'Mínimo\n{min_data}',
    (min_idx, min_y),
    textcoords='offset points',
    xytext=(0, -50),
    ha='center',
    fontsize=9,
    fontweight='bold',
    color=COR_MIN,
    bbox=dict(
        boxstyle='round,pad=0.3',
        fc='white',
        ec=COR_MIN,
        alpha=0.95
    )
)

# Formata eixo Y como percentual
ax.yaxis.set_major_formatter(
    FuncFormatter(lambda y, _: f'{y:.1f}%')
)

# Datas
ax.set_xticks(range(len(cancelamento_mensal)))
ax.set_xticklabels(
    cancelamento_mensal['ano_mes_formatado'],
    rotation=45,
    ha='right',
    fontsize=9
)

# Título
ax.set_title(
    'Evolução da Taxa de Cancelamento',
    fontsize=16,
    fontweight='bold',
    pad=20
)

# Eixos
ax.set_xlabel('')
ax.set_ylabel(
    'Taxa de Cancelamento (%)',
    fontsize=12
)

# Remove grades
ax.grid(False)

# Remove bordas superior e direita
sns.despine()

# Espaço para as anotações
y_min = cancelamento_mensal['taxa_cancelamento'].min()
y_max = cancelamento_mensal['taxa_cancelamento'].max()

ax.set_ylim(
    max(0, y_min - 0.20),
    y_max + 0.35
)

plt.tight_layout()
plt.show()
```

```python
cliente_satisfacao = (
    orders
    .merge(
        customers[['customer_id', 'customer_unique_id']],
        on='customer_id'
    )
    .merge(
        order_reviews[['order_id', 'review_score']],
        on='order_id'
    )
)

cliente_satisfacao = (
    cliente_satisfacao
    .groupby('customer_unique_id')
    .agg(
        satisfacao_media=('review_score', 'mean'),
        total_compras=('order_id', 'count')
    )
    .reset_index()
)

cliente_satisfacao.head()
```

```python
cliente_satisfacao['recomprou'] = np.where(
    cliente_satisfacao['total_compras'] > 1,
    1,
    0
)
```

```python
cliente_satisfacao['faixa_satisfacao'] = pd.cut(
    cliente_satisfacao['satisfacao_media'],
    bins=[0, 2, 3, 4, 5],
    labels=[
        'Baixa (1-2)',
        'Média (2-3)',
        'Boa (3-4)',
        'Alta (4-5)'
    ]
)
```

```python
recompra_satisfacao = (
    cliente_satisfacao
    .groupby('faixa_satisfacao', observed=False)
    .agg(
        clientes=('customer_unique_id', 'count'),
        taxa_recompra=('recomprou', 'mean')
    )
    .reset_index()
)

recompra_satisfacao['taxa_recompra'] = (
    recompra_satisfacao['taxa_recompra'] * 100
)

display(recompra_satisfacao)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Estilo
sns.set_theme(style="white")

fig, ax = plt.subplots(figsize=(10, 5))

# Linha principal
sns.lineplot(
    data=recompra_satisfacao,
    x='faixa_satisfacao',
    y='taxa_recompra',
    marker='o',
    markersize=10,
    linewidth=3,
    color='#1F4E79',
    ax=ax
)

# Anotações
for x, y in zip(
    recompra_satisfacao['faixa_satisfacao'],
    recompra_satisfacao['taxa_recompra']
):
    ax.annotate(
        f'{y:.1f}%',
        (x, y),
        xytext=(0, 12),
        textcoords='offset points',
        ha='center',
        fontsize=10,
        fontweight='bold',
        bbox=dict(
            boxstyle='round,pad=0.3',
            fc='white',
            ec='#1F4E79',
            alpha=0.9
        )
    )

# Títulos
ax.set_title(
    'Relação entre Satisfação e Taxa de Recompra',
    fontsize=18,
    fontweight='bold',
    pad=20
)

ax.set_xlabel('Faixa de Satisfação', fontsize=12)
ax.set_ylabel('Taxa de Recompra (%)', fontsize=12)

# Ajuste do eixo Y
ymin = recompra_satisfacao['taxa_recompra'].min()
ymax = recompra_satisfacao['taxa_recompra'].max()

ax.set_ylim(ymin - 2, ymax + 5)

# Personalização
ax.tick_params(axis='both', labelsize=11)


sns.despine(left=True, bottom=False)

plt.tight_layout()
plt.show()
```

A análise indica que a satisfação do cliente possui relação com a taxa de recompra, mas essa associação não é linear. Clientes com satisfação média (2–3) apresentam a maior taxa de recompra (5,2%), enquanto aqueles com baixa satisfação (1–2) registram a menor taxa (1,9%). Já os grupos com avaliações boas (3–4) e altas (4–5) apresentam taxas intermediárias, de 3,0% e 3,3%, respectivamente.

Esse comportamento sugere que a recompra não depende exclusivamente da satisfação expressa nas avaliações. Outros fatores, como necessidade recorrente do produto, preço, promoções e conveniência da plataforma, também podem influenciar a decisão de compra. Ainda assim, a baixa taxa de recompra entre clientes menos satisfeitos reforça a importância de monitorar a experiência do consumidor para reduzir a perda de clientes e estimular a fidelização.
